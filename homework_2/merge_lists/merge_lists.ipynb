## Merge lists

Даны два отсортированных односвязных списка list1 и list2

Необходимо объединить их в один новый отсортированный список.
Новый список должен быть составлен слиянием узлов двух исходных списков.
Вернуть необходимо голову объединенного списка.
 

Вход: list1 = [1,2,4], list2 = [1,3,4]

Выход: [1,1,2,3,4,4]

Требования:
* реализовать два способа решения поставленной задачи: с использованием фиктивного элемента и без него
* написать тесты

Формат сдачи: в своем репозитории /homework_2/merge_lists/

```python
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None
```

## С использованием фиктивного элемента

```python
def merge_with_dummy(list1, list2):
    # создаем фиктивный узел
    # он нужен только как удобное начало нового списка
    dummy = Node(0)

    # tail всегда указывает на последний элемент результата
    tail = dummy

    # пока оба списка не закончились
    while list1 is not None and list2 is not None:

        # сравниваем текущие элементы двух списков
        if list1.value <= list2.value:
            # присоединяем элемент первого списка
            tail.next = list1

            # двигаемся дальше по первому списку
            list1 = list1.next

        else:
            # присоединяем элемент второго списка
            tail.next = list2

            # двигаемся дальше по второму списку
            list2 = list2.next

        # передвигаем конец результата
        tail = tail.next

    # если один из списков закончился,
    # присоединяем оставшуюся часть другого списка
    if list1 is not None:
        tail.next = list1
    else:
        tail.next = list2

    # dummy в ответ не входит
    return dummy.next
```

```python
list1 = Node(1)
list1.next = Node(2)
list1.next.next = Node(4)

list2 = Node(1)
list2.next = Node(3)
list2.next.next = Node(4)

result = merge_with_dummy(list1, list2)

current = result

while current is not None:
    print(current.value, end=" ")
    current = current.next
```

## Без использования фиктивного элемента

```python
def merge_without_dummy(list1, list2):
    # если первый список пустой,
    # результатом сразу будет второй
    if list1 is None:
        return list2

    # если второй список пустой,
    # результатом сразу будет первый
    if list2 is None:
        return list1

    # выбираем первый элемент нового списка
    if list1.value <= list2.value:
        head = list1
        list1 = list1.next
    else:
        head = list2
        list2 = list2.next

    # tail указывает на последний элемент результата
    tail = head

    # дальше сравниваем элементы двух списков
    while list1 is not None and list2 is not None:

        if list1.value <= list2.value:
            tail.next = list1
            list1 = list1.next
        else:
            tail.next = list2
            list2 = list2.next

        tail = tail.next

    # присоединяем оставшуюся часть
    if list1 is not None:
        tail.next = list1
    else:
        tail.next = list2

    return head
```

```python
list1 = Node(1)
list1.next = Node(2)
list1.next.next = Node(4)

list2 = Node(1)
list2.next = Node(3)
list2.next.next = Node(4)

result = merge_without_dummy(list1, list2)

current = result

while current is not None:
    print(current.value, end=" ")
    current = current.next
```

## Тесты

```python
import unittest
import sys


class TestMergeLists(unittest.TestCase):

    def test_with_dummy(self):
        # list1 = 1 -> 2 -> 4
        list1 = Node(1)
        list1.next = Node(2)
        list1.next.next = Node(4)

        # list2 = 1 -> 3 -> 4
        list2 = Node(1)
        list2.next = Node(3)
        list2.next.next = Node(4)

        result = merge_with_dummy(list1, list2)

        # проверяем значения по порядку
        self.assertEqual(result.value, 1)
        self.assertEqual(result.next.value, 1)
        self.assertEqual(result.next.next.value, 2)
        self.assertEqual(result.next.next.next.value, 3)
        self.assertEqual(result.next.next.next.next.value, 4)
        self.assertEqual(result.next.next.next.next.next.value, 4)

        # после последнего элемента список должен закончиться
        self.assertIsNone(result.next.next.next.next.next.next)

    def test_without_dummy(self):
        list1 = Node(1)
        list1.next = Node(2)
        list1.next.next = Node(4)

        list2 = Node(1)
        list2.next = Node(3)
        list2.next.next = Node(4)

        result = merge_without_dummy(list1, list2)

        self.assertEqual(result.value, 1)
        self.assertEqual(result.next.value, 1)
        self.assertEqual(result.next.next.value, 2)
        self.assertEqual(result.next.next.next.value, 3)
        self.assertEqual(result.next.next.next.next.value, 4)
        self.assertEqual(result.next.next.next.next.next.value, 4)

        self.assertIsNone(result.next.next.next.next.next.next)

    def test_first_list_empty(self):
        list1 = None

        list2 = Node(1)
        list2.next = Node(2)

        result = merge_without_dummy(list1, list2)

        self.assertEqual(result.value, 1)
        self.assertEqual(result.next.value, 2)
        self.assertIsNone(result.next.next)

    def test_second_list_empty(self):
        list1 = Node(1)
        list1.next = Node(2)

        list2 = None

        result = merge_with_dummy(list1, list2)

        self.assertEqual(result.value, 1)
        self.assertEqual(result.next.value, 2)
        self.assertIsNone(result.next.next)

    def test_both_lists_empty(self):
        result = merge_with_dummy(None, None)

        self.assertIsNone(result)

    def test_one_element(self):
        list1 = Node(1)
        list2 = Node(2)

        result = merge_without_dummy(list1, list2)

        self.assertEqual(result.value, 1)
        self.assertEqual(result.next.value, 2)
        self.assertIsNone(result.next.next)


suite = unittest.TestLoader().loadTestsFromTestCase(TestMergeLists)

runner = unittest.TextTestRunner(
    stream=sys.stdout,
    verbosity=2
)

result = runner.run(suite)
```

## Оценка сложности

Пусть в первом списке `n` элементов, а во втором — `m`.

### Время: O(n + m)

На каждом шаге мы сравниваем текущие элементы двух списков и передвигаемся вперед по одному из них.

Каждый узел первого списка обрабатывается не более одного раза.
Каждый узел второго списка тоже обрабатывается не более одного раза.

То есть максимум мы пройдем:

`n + m`

узлов.

Поэтому временная сложность обоих способов:

**O(n + m)**

Например, если:

- в `list1` 3 элемента;
- в `list2` 3 элемента,

то всего нужно обработать не более 6 узлов.

### Память: O(1)

Новый список из новых узлов мы не создаем.

Мы используем уже существующие узлы из `list1` и `list2` и просто меняем их ссылки `next`.

Используется только несколько дополнительных переменных:

- `list1`;
- `list2`;
- `tail`;
- `head` или `dummy`.

Количество этих переменных не зависит от размера списков.

Поэтому дополнительная память:

**O(1)**

В способе с `dummy` создается только один дополнительный узел, поэтому сложность по памяти все равно остается `O(1)`.

### Как работает способ с dummy

В начале создаем фиктивный узел:

`dummy → 0`

`tail` сначала указывает на него:

`tail`
`↓`
`dummy → 0`

Дальше к `tail.next` по очереди присоединяем подходящие узлы.

После первого шага:

`dummy → 1`

После второго:

`dummy → 1 → 1`

Дальше:

`dummy → 1 → 1 → 2 → 3 → 4 → 4`

Фиктивный узел `dummy` нужен только для удобного построения списка.

Он не входит в результат.

Поэтому возвращаем:

`dummy.next`

То есть настоящую голову списка:

`1 → 1 → 2 → 3 → 4 → 4`

### Как работает способ без dummy

Без фиктивного элемента сначала нужно отдельно выбрать голову нового списка.

Даны:

`list1 = 1 → 2 → 4`

`list2 = 1 → 3 → 4`

Сравниваем первые элементы:

`1 <= 1`

Поэтому первый узел `list1` становится `head`:

`head`
`↓`
`1`

После этого `tail` тоже указывает на этот элемент:

`head`
`↓`
`1`
`↑`
`tail`

Дальше алгоритм работает так же, как вариант с `dummy`:
сравниваем текущие элементы двух списков и присоединяем меньший.

В итоге:

`head`
`↓`
`1 → 1 → 2 → 3 → 4 → 4`

Главное отличие от способа с `dummy` в том, что первый элемент результата приходится выбирать отдельно.
