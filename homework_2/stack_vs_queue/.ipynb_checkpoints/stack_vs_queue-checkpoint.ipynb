## Stack vs queue

## Объяснение решения

Стек и очередь реализованы на основе односвязного списка.

Каждый элемент связного списка это `Node`, который хранит:
- значение `value`;
- ссылку `next` на следующий элемент.

```python
#один элемент связного списка 

class Node:
    def __init__(self, value):
        # значение узла
        self.value = value

        # ссылка на следующий узел
        self.next = None
```

### Stack

Стек работает по принципу LIFO: последний добавленный элемент удаляется первым.

При `push` новый элемент становится верхушкой стека и ссылается на предыдущую верхушку.

При `pop` верхний элемент удаляется, а следующий за ним элемент становится новой верхушкой.


```python
#реализуем стек через связный список
#LIFO

class Stack:
    
    def __init__(self):
        # верхушка стека
        self.top = None
        # ссылка на следующий узел
        self.size = 0

    #добавление элемента в стэк
    def push(self, value):

        # создаем новый узел
        new_node = Node(value)

        # текущая верхушка становится следующей
        new_node.next = self.top

        # новый узел становится верхушкой
        self.top = new_node

        self.size += 1

    #достать и удалить элемент из стека
    def pop(self):
        
        # если стек пустой
        if self.top is None:
            raise IndexError("Стек пуст")

        # запоминаем значение верхушки, чтобы ее вернуть потом
        value = self.top.value

        # следующий элемент становится новой верхушкой
        self.top = self.top.next

        self.size -= 1

        return value
    
    def peek(self):
        # возвращаем верхний элемент без удаления
        if self.top is None:
            raise IndexError("Стек пуст")

        return self.top.value

    def is_empty(self):
        return self.top is None

    def __len__(self):
        return self.size
```

## Визуализация работы Stack

Добавим элементы:

`push(1)`

Получаем:

`top → [1] → None`

Затем:

`push(2)`

Получаем:

`top → [2] → [1] → None`

Затем:

`push(3)`

Получаем:

`top → [3] → [2] → [1] → None`

Теперь вызываем:

`pop()`

Удаляется верхний элемент `3`.

Остается:

`top → [2] → [1] → None`

То есть стек работает по принципу:

**последний пришел — первый вышел**

### Queue

Очередь работает по принципу FIFO: первый добавленный элемент удаляется первым.

Для очереди хранятся две ссылки:
- `head` — первый элемент;
- `tail` — последний элемент.

При `enqueue` новый элемент добавляется после `tail`.

При `dequeue` удаляется элемент `head`, а следующий элемент становится новым `head`.


```python
#реализуем очередь через свзяный список
#FIFO

class Queue:
    
    def __init__(self):
        self.head = None
        self.tail = None
        self.size = 0

    #добавляем элемент в очередь
    def enqueue(self, value):
        new_node = Node(value)

        if self.tail is None:
            self.head = new_node
            self.tail = new_node
        else:
            self.tail.next = new_node
            self.tail = new_node

        self.size += 1

    #удаляем элемент из очереди
    def dequeue(self):
        if self.head is None:
            raise IndexError("Очередь пуста")

        value = self.head.value
        self.head = self.head.next

        if self.head is None:
            self.tail = None

        self.size -= 1

        return value

    # смотрим первый элемент, но не удаляем его
    def peek(self):
        if self.head is None:
            raise IndexError("Очередь пуста")

        return self.head.value

    def is_empty(self):
        return self.head is None

    def __len__(self):
        return self.size
```

## Визуализация работы Queue

Добавим элементы:

`enqueue(1)`

`head → [1] ← tail`

Затем:

`enqueue(2)`

`head → [1] → [2] ← tail`

Затем:

`enqueue(3)`

`head → [1] → [2] → [3] ← tail`

Теперь вызываем:

`dequeue()`

Удаляется первый элемент `1`.

Получаем:

`head → [2] → [3] ← tail`

То есть очередь работает по принципу:

**первый пришел — первый вышел.**

## Оценка сложности

### Stack

`push` — `O(1)`

Новый элемент сразу добавляется в начало связного списка.

`pop` — `O(1)`

Удаляется только верхний элемент.

`peek` — `O(1)`

Мы сразу имеем ссылку на верхушку стека.

### Queue

`enqueue` — `O(1)`

Мы храним ссылку `tail` на последний элемент, поэтому не нужно проходить всю очередь.

`dequeue` — `O(1)`

Удаляется первый элемент, на который указывает `head`.

`peek` — `O(1)`.

### Память

Для хранения `n` элементов требуется `O(n)` памяти.

Для каждого значения создается отдельный узел `Node`.

```python
import unittest
import sys


class TestStack(unittest.TestCase):

    def test_push_and_pop(self):
        stack = Stack()

        stack.push(1)
        stack.push(2)
        stack.push(3)

        # стек работает по LIFO
        self.assertEqual(stack.pop(), 3)
        self.assertEqual(stack.pop(), 2)
        self.assertEqual(stack.pop(), 1)

    def test_peek(self):
        stack = Stack()

        stack.push(10)
        stack.push(20)

        # peek показывает верхушку, но не удаляет ее
        self.assertEqual(stack.peek(), 20)
        self.assertEqual(len(stack), 2)

    def test_size(self):
        stack = Stack()

        self.assertEqual(len(stack), 0)

        stack.push(1)
        stack.push(2)

        self.assertEqual(len(stack), 2)

        stack.pop()

        self.assertEqual(len(stack), 1)

    def test_is_empty(self):
        stack = Stack()

        self.assertTrue(stack.is_empty())

        stack.push(1)

        self.assertFalse(stack.is_empty())

    def test_one_element(self):
        stack = Stack()

        stack.push(5)

        self.assertEqual(stack.pop(), 5)
        self.assertTrue(stack.is_empty())

    def test_pop_empty_stack(self):
        stack = Stack()

        # проверяем ошибку при удалении из пустого стека
        with self.assertRaises(IndexError):
            stack.pop()


class TestQueue(unittest.TestCase):

    def test_enqueue_and_dequeue(self):
        queue = Queue()

        queue.enqueue(1)
        queue.enqueue(2)
        queue.enqueue(3)

        # очередь работает по FIFO
        self.assertEqual(queue.dequeue(), 1)
        self.assertEqual(queue.dequeue(), 2)
        self.assertEqual(queue.dequeue(), 3)

    def test_peek(self):
        queue = Queue()

        queue.enqueue(10)
        queue.enqueue(20)

        # peek показывает первый элемент, но не удаляет его
        self.assertEqual(queue.peek(), 10)
        self.assertEqual(len(queue), 2)

    def test_size(self):
        queue = Queue()

        self.assertEqual(len(queue), 0)

        queue.enqueue(1)
        queue.enqueue(2)

        self.assertEqual(len(queue), 2)

        queue.dequeue()

        self.assertEqual(len(queue), 1)

    def test_is_empty(self):
        queue = Queue()

        self.assertTrue(queue.is_empty())

        queue.enqueue(1)

        self.assertFalse(queue.is_empty())

    def test_one_element(self):
        queue = Queue()

        queue.enqueue(5)

        self.assertEqual(queue.dequeue(), 5)
        self.assertTrue(queue.is_empty())

    def test_dequeue_empty_queue(self):
        queue = Queue()

        # проверяем ошибку при удалении из пустой очереди
        with self.assertRaises(IndexError):
            queue.dequeue()


suite = unittest.TestSuite()

suite.addTests(
    unittest.TestLoader().loadTestsFromTestCase(TestStack)
)

suite.addTests(
    unittest.TestLoader().loadTestsFromTestCase(TestQueue)
)

runner = unittest.TextTestRunner(
    stream=sys.stdout,
    verbosity=2
)

result = runner.run(suite)
```
