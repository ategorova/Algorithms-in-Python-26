### Задача 2. Сумма
На вход подается массив целых положительных чисел, разделенных пробелом.

Нужно найти максимальную сумму элементов массива, которая делится на 2.

Требования:
* Напишите тесты, которые проверяют функцию на разных случаях.
* Реализовать тесты можно любым способом (unittest, pytest)


Пример 1:

Вход: 5 7 13 2 14

Выход: 36

Пояснение: Берем 2, 7, 13, 14              

Пример 2:

Вход: 3

Выход: 0

Пояснение: 3 не делится на 2, поэтому в ответе 0.



```python
def max_even_sum(numbers: list[int]) -> int:
    #изначально лучше, чем суммы всех числел, быть не может
    #но если она не делится на два, тогда нам надо убрать минимальное нечетное число
    total = sum(numbers)

    if total % 2 == 0:
        return total
    #ищем минимальное число, если сумма не делится на два сразу
    min_odd = min(number for number in numbers if number % 2 != 0) #в худшем случае нам все равно надо перебрать все нечетные элементы
    #сортировка была бы хуже, так что лучше перебрать

    return total - min_odd
```

```python
import unittest
import sys


class TestMaxEvenSum(unittest.TestCase):

    def test_example(self):
        self.assertEqual(max_even_sum([5, 7, 13, 2, 14]), 36)

    def test_one_odd_number(self):
        self.assertEqual(max_even_sum([3]), 0)

    def test_one_even_number(self):
        self.assertEqual(max_even_sum([8]), 8)

    def test_all_even_numbers(self):
        self.assertEqual(max_even_sum([2, 4, 6, 8]), 20)

    def test_even_sum_with_odd_numbers(self):
        self.assertEqual(max_even_sum([1, 3, 5, 7]), 16)

    def test_remove_smallest_odd(self):
        self.assertEqual(max_even_sum([1, 2, 4]), 6)

    def test_many_odd_numbers(self):
        self.assertEqual(max_even_sum([9, 3, 7]), 16)

    def test_empty_array(self):
        self.assertEqual(max_even_sum([]), 0)


suite = unittest.TestLoader().loadTestsFromTestCase(TestMaxEvenSum)
runner = unittest.TextTestRunner(stream=sys.stdout, verbosity=2)
runner.run(suite);
```

### Оценка сложности

**время: O(n)**

алгоритм один раз проходит по массиву из `n` элементов

на каждой итерации выполняется постоянное количество операций:
- прибавление элемента к общей сумме;
- проверка его четности;
- при необходимости сравнение с минимальным нечетным элементом.

Поэтому временная сложность равна `O(n)`

**память: O(1)**

алгоритм не создает дополнительных массивов или других структур данных,
размер которых зависит от количества элементов

используются только переменные `total` и `min_odd`,
поэтому дополнительная память равна `O(1)`

### Визуализация работы алгоритма

рассмотрим массив:

`[5, 7, 13, 2, 14]`

во время прохода по массиву считаем общую сумму и ищем минимальное нечетное число.

| Шаг | Число | total | min_odd |
|---|---:|---:|---:|
| начало | — | 0 | None |
| 1 | 5 | 5 | 5 |
| 2 | 7 | 12 | 5 |
| 3 | 13 | 25 | 5 |
| 4 | 2 | 27 | 5 |
| 5 | 14 | 41 | 5 |

после прохода получили:

`total = 41`

`min_odd = 5`

сумма `41` нечетная, поэтому вычитаем минимальное нечетное число:

`41 - 5 = 36`

ответ: `36`
