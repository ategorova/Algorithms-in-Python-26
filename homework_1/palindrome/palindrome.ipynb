### Задача 1. Палиндром

Напишите программу, которая проверяет, является ли целое положительное число палиндромом.
Сделайте это без использования строк.

#### Требования:
* Напишите тесты, которые проверяют функцию на разных случаях.
* Реализовать тесты можно любым способом (unittest, pytest)


Пример 1:

Вход: 121

Выход: True                 

Пример 2:

Вход: 31

Выход: False

```python
def is_palindrome(n: int) -> bool:
    # если передали не целое число, сразу возвращаем False
    if type(n) is not int:
        return False

    # если число не положительное, сразу возвращаем False
    if n <= 0:
        return False

    # сохраняем исходное число
    # и создаем перевернутое число, пока равное нулю
    original = n
    reversed_number = 0

    # пока n больше нуля, по одной цифре забираем число справа
    while n > 0:
        digit = n % 10  # берем последнюю цифру

        # добавляем эту цифру в конец перевернутого числа
        reversed_number = reversed_number * 10 + digit

        # убираем последнюю уже обработанную цифру из n
        n //= 10

    # проверяем, совпадает ли исходное число с перевернутым
    return original == reversed_number
```

```python
if __name__ == "__main__":
    try:
        number = int(input("Введите положительное целое число: "))
        print(is_palindrome(number))
    except ValueError:
        print("Ошибка: необходимо ввести целое число")
```

Вообще 0 является палиндромом, но в задаче написано, что целое положительное число берем

```python
import unittest


class TestPalindrome(unittest.TestCase):

    def test_palindrome(self):
        self.assertTrue(is_palindrome(121))

    def test_not_palindrome(self):
        self.assertFalse(is_palindrome(31))

    def test_one_digit(self):
        self.assertTrue(is_palindrome(7))

    def test_even_length_palindrome(self):
        self.assertTrue(is_palindrome(1221))

    def test_palindrome_with_zero_inside(self):
        self.assertTrue(is_palindrome(1001))

    def test_negative_number(self):
        self.assertFalse(is_palindrome(-121))

    def test_zero(self):
        self.assertFalse(is_palindrome(0))

    def test_float(self):
        self.assertFalse(is_palindrome(12.5))

    def test_string(self):
        self.assertFalse(is_palindrome("121"))


import sys

suite = unittest.TestLoader().loadTestsFromTestCase(TestPalindrome)
runner = unittest.TextTestRunner(stream=sys.stdout, verbosity=2)
result = runner.run(suite)
```

#### оценка сложности

**Время: O(log n)**

на каждой итерации цикла число `n` делится на 10:

`12321 → 1232 → 123 → 12 → 1 → 0`

то есть за одну итерацию убирается одна цифра
колво цифр в числе `n` примерно равно `log₁₀(n)`, поэтому цикл выполняется `O(log n)` раз.

**память: O(1)**

алгоритм не создает массивов или других структур, размер которых зависит от числа
используется только несколько переменных: `original`, `reversed_number` и `digit`, поэтому дополнительная память постоянна

#### визуализация работы алгоритма

рассмотрим число `12321`

изначально:

- `original = 12321`
- `reversed_number = 0`

на каждом шаге берем последнюю цифру числа с помощью `% 10`,
добавляем ее в `reversed_number`, а затем удаляем эту цифру из `n` с помощью `// 10`

| Шаг | n до обработки | digit = n % 10 | reversed_number | n после n // 10 |
|---|---:|---:|---:|---:|
| 1 | 12321 | 1 | 1 | 1232 |
| 2 | 1232 | 2 | 12 | 123 |
| 3 | 123 | 3 | 123 | 12 |
| 4 | 12 | 2 | 1232 | 1 |
| 5 | 1 | 1 | 12321 | 0 |

после завершения цикла:

`original = 12321`

`reversed_number = 12321`

числа равны, поэтому функция возвращает `True`
