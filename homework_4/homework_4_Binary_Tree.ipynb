# Homework 4. Binary Tree

В работе реализованы три задачи:

1. **Tree Traversal** - BST, вставка и шесть обходов
2. **Validate BST** - проверка произвольного бинарного дерева
3. **Balanced Binary Tree** - проверка баланса по высоте

Во всех задачах используется один класс `Node`.

В этой реализации BST строгий:
- слева значения меньше текущего узла;
- справа значения больше;
- дубликаты при вставке не добавляются.

## Общий класс узла

Каждый узел хранит значение и ссылки на левого и правого потомка.

````python
class Node:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None
````

# Задача 1. Tree Traversal

## Идея

В BST значения меньше текущего узла идут влево, больше — вправо.

Для примера вставим:

`8, 3, 10, 1, 6, 14, 4, 7, 13`

Получим:

```text
          8
        /   \
       3     10
      / \      \
     1   6      14
        / \     /
       4   7   13
```

Обходы отличаются только порядком посещения: корень, левое и правое поддерево.

````python
class BinarySearchTree:
    def __init__(self):
        self.root = None

    def insert(self, value):
        # если дерево пустое — создаем корень
        if self.root is None:
            self.root = Node(value)
            return

        current = self.root

        while True:
            # меньше — идем влево
            if value < current.value:
                if current.left is None:
                    current.left = Node(value)
                    return
                current = current.left

            # больше — идем вправо
            elif value > current.value:
                if current.right is None:
                    current.right = Node(value)
                    return
                current = current.right

            # дубликаты не добавляем
            else:
                return

    def preorder(self):
        result = []

        def traverse(node):
            if node is None:
                return
            result.append(node.value)   # корень
            traverse(node.left)        # левое
            traverse(node.right)       # правое

        traverse(self.root)
        return result

    def inorder(self):
        result = []

        def traverse(node):
            if node is None:
                return
            traverse(node.left)        # левое
            result.append(node.value)  # корень
            traverse(node.right)       # правое

        traverse(self.root)
        return result

    def postorder(self):
        result = []

        def traverse(node):
            if node is None:
                return
            traverse(node.left)        # левое
            traverse(node.right)       # правое
            result.append(node.value)  # корень

        traverse(self.root)
        return result

    def reverse_preorder(self):
        result = []

        def traverse(node):
            if node is None:
                return
            result.append(node.value)   # корень
            traverse(node.right)       # правое
            traverse(node.left)        # левое

        traverse(self.root)
        return result

    def reverse_inorder(self):
        result = []

        def traverse(node):
            if node is None:
                return
            traverse(node.right)       # правое
            result.append(node.value)  # корень
            traverse(node.left)        # левое

        traverse(self.root)
        return result

    def reverse_postorder(self):
        result = []

        def traverse(node):
            if node is None:
                return
            traverse(node.right)       # правое
            traverse(node.left)        # левое
            result.append(node.value)  # корень

        traverse(self.root)
        return result
````

````python
tree = BinarySearchTree()

for value in [8, 3, 10, 1, 6, 14, 4, 7, 13]:
    tree.insert(value)

print("Preorder:", tree.preorder())
print("Inorder:", tree.inorder())
print("Postorder:", tree.postorder())
print("Reverse preorder:", tree.reverse_preorder())
print("Reverse inorder:", tree.reverse_inorder())
print("Reverse postorder:", tree.reverse_postorder())
````

## Визуализация обходов

Для дерева:

```text
          8
        /   \
       3     10
      / \      \
     1   6      14
        / \     /
       4   7   13
```

### Preorder: корень → левое → правое

`8 → 3 → 1 → 6 → 4 → 7 → 10 → 14 → 13`

### Inorder: левое → корень → правое

`1 → 3 → 4 → 6 → 7 → 8 → 10 → 13 → 14`

Для BST этот обход дает значения по возрастанию.

### Postorder: левое → правое → корень

`1 → 4 → 7 → 6 → 3 → 13 → 14 → 10 → 8`

### Reverse preorder: корень → правое → левое

`8 → 10 → 14 → 13 → 3 → 6 → 7 → 4 → 1`

### Reverse inorder: правое → корень → левое

`14 → 13 → 10 → 8 → 7 → 6 → 4 → 3 → 1`

Для BST этот обход дает значения по убыванию.

### Reverse postorder: правое → левое → корень

`13 → 14 → 10 → 7 → 4 → 6 → 1 → 3 → 8`

## Сложность задачи 1

Пусть `n` — число узлов, `h` — высота дерева.

### Вставка

Одна вставка идет от корня вниз по одному пути:

- в среднем: `O(log n)` для достаточно сбалансированного дерева;
- в худшем случае: `O(n)` для вырожденного дерева.

Например, при вставке `1, 2, 3, 4, 5` получится цепочка, и для нового элемента придется пройти почти все узлы.

### Обходы

Каждый узел посещается один раз.

**Время: `O(n)`**

Рекурсивный стек хранит путь от корня до текущего узла.

**Память: `O(h)`**

- сбалансированное дерево: `O(log n)`;
- худший случай: `O(n)`.

# Задача 2. Validate BST

## Идея

Проверки только прямых детей недостаточно.

Например:

```text
       10
      /  \
     5    15
         /
        6
```

`6 < 15`, но `6` находится в правом поддереве `10`, поэтому обязано быть больше `10`.

Поэтому для каждого узла храним допустимый диапазон:

`lower < node.value < upper`

При движении влево уменьшаем верхнюю границу, при движении вправо увеличиваем нижнюю.

````python
def is_valid_bst(root):
    def validate(node, lower, upper):
        # пустое поддерево корректно
        if node is None:
            return True

        # значение должно лежать строго внутри допустимого диапазона
        if not (lower < node.value < upper):
            return False

        # слева все значения должны быть меньше текущего
        if not validate(node.left, lower, node.value):
            return False

        # справа все значения должны быть больше текущего
        return validate(node.right, node.value, upper)

    return validate(root, float("-inf"), float("inf"))
````

## Визуализация Validate BST

### Корректное дерево

```text
        8
      /   \
     3     10
    / \      \
   1   6      14
```

Проверяем диапазоны:

| Узел | Допустимый диапазон | Результат |
|---:|---|---|
| 8 | `(-∞, +∞)` | подходит |
| 3 | `(-∞, 8)` | подходит |
| 1 | `(-∞, 3)` | подходит |
| 6 | `(3, 8)` | подходит |
| 10 | `(8, +∞)` | подходит |
| 14 | `(10, +∞)` | подходит |

Ответ: `True`.

### Некорректное дерево

```text
       10
      /  \
     5    15
         /
        6
```

Для `6` допустимый диапазон — `(10, 15)`.

Но `6` в него не попадает.

Ответ: `False`.

## Сложность задачи 2

Каждый узел проверяется один раз.

**Время: `O(n)`**

В рекурсивном стеке хранится путь от корня до текущего узла.

**Память: `O(h)`**

- сбалансированное дерево: `O(log n)`;
- вырожденное дерево: `O(n)`.

# Задача 3. Balanced Binary Tree

## Идея

Для каждого узла сравниваем высоты левого и правого поддеревьев.

Дерево сбалансировано, если для каждого узла:

`|left_height - right_height| <= 1`

Чтобы не пересчитывать высоты много раз, делаем один рекурсивный проход.

Функция `height` возвращает:

- высоту поддерева, если оно сбалансировано;
- `-1`, если уже найден дисбаланс.

Как только получили `-1`, можно сразу передавать его выше и прекращать лишние вычисления.

````python
def is_balanced(root):
    def height(node):
        # высота пустого дерева равна 0
        if node is None:
            return 0

        left_height = height(node.left)

        # если слева уже нашли дисбаланс
        if left_height == -1:
            return -1

        right_height = height(node.right)

        # если справа уже нашли дисбаланс
        if right_height == -1:
            return -1

        # проверяем текущий узел
        if abs(left_height - right_height) > 1:
            return -1

        # высота текущего поддерева
        return 1 + max(left_height, right_height)

    return height(root) != -1
````

## Визуализация Balanced Binary Tree

### Сбалансированное дерево

```text
        8
      /   \
     4     12
    / \    / \
   2   6  10  14
```

Листья имеют высоту 1.

Для `4`:

- слева высота 1;
- справа высота 1;
- разница `0`.

Для `12` то же самое.

Для корня `8`:

- слева высота 2;
- справа высота 2;
- разница `0`.

Ответ: `True`.

### Несбалансированное дерево

```text
        8
       /
      4
     /
    2
   /
  1
```

Считаем снизу вверх:

| Узел | Высота слева | Высота справа | Разница |
|---:|---:|---:|---:|
| 1 | 0 | 0 | 0 |
| 2 | 1 | 0 | 1 |
| 4 | 2 | 0 | **2** |

На узле `4` разница больше 1, поэтому возвращаем `-1`.

Ответ: `False`.

## Сложность задачи 3

Каждый узел посещается не более одного раза.

**Время: `O(n)`**

Это лучше наивного решения, где высота одних и тех же поддеревьев пересчитывается много раз и можно получить `O(n²)`.

Рекурсивный стек содержит путь от корня до текущего узла.

**Память: `O(h)`**

- сбалансированное дерево: `O(log n)`;
- вырожденное дерево: `O(n)`.

# Тесты

Проверяем обычные и граничные случаи для всех трех задач.

````python
import unittest
import sys


class TestTreeTraversal(unittest.TestCase):
    def setUp(self):
        self.tree = BinarySearchTree()
        for value in [8, 3, 10, 1, 6, 14, 4, 7, 13]:
            self.tree.insert(value)

    def test_preorder(self):
        self.assertEqual(
            self.tree.preorder(),
            [8, 3, 1, 6, 4, 7, 10, 14, 13]
        )

    def test_inorder(self):
        self.assertEqual(
            self.tree.inorder(),
            [1, 3, 4, 6, 7, 8, 10, 13, 14]
        )

    def test_postorder(self):
        self.assertEqual(
            self.tree.postorder(),
            [1, 4, 7, 6, 3, 13, 14, 10, 8]
        )

    def test_reverse_preorder(self):
        self.assertEqual(
            self.tree.reverse_preorder(),
            [8, 10, 14, 13, 3, 6, 7, 4, 1]
        )

    def test_reverse_inorder(self):
        self.assertEqual(
            self.tree.reverse_inorder(),
            [14, 13, 10, 8, 7, 6, 4, 3, 1]
        )

    def test_reverse_postorder(self):
        self.assertEqual(
            self.tree.reverse_postorder(),
            [13, 14, 10, 7, 4, 6, 1, 3, 8]
        )

    def test_empty_tree(self):
        tree = BinarySearchTree()
        self.assertEqual(tree.preorder(), [])
        self.assertEqual(tree.inorder(), [])
        self.assertEqual(tree.postorder(), [])
        self.assertEqual(tree.reverse_preorder(), [])
        self.assertEqual(tree.reverse_inorder(), [])
        self.assertEqual(tree.reverse_postorder(), [])

    def test_one_node(self):
        tree = BinarySearchTree()
        tree.insert(5)
        self.assertEqual(tree.inorder(), [5])

    def test_duplicate_is_not_added(self):
        tree = BinarySearchTree()
        tree.insert(5)
        tree.insert(5)
        self.assertEqual(tree.inorder(), [5])


class TestValidateBST(unittest.TestCase):
    def test_valid_bst(self):
        root = Node(8)
        root.left = Node(3)
        root.right = Node(10)
        root.left.left = Node(1)
        root.left.right = Node(6)
        root.right.right = Node(14)
        self.assertTrue(is_valid_bst(root))

    def test_invalid_bst_deep_error(self):
        root = Node(10)
        root.left = Node(5)
        root.right = Node(15)
        root.right.left = Node(6)
        self.assertFalse(is_valid_bst(root))

    def test_invalid_direct_child(self):
        root = Node(10)
        root.left = Node(12)
        self.assertFalse(is_valid_bst(root))

    def test_empty_tree(self):
        self.assertTrue(is_valid_bst(None))

    def test_one_node(self):
        self.assertTrue(is_valid_bst(Node(5)))

    def test_duplicate_value(self):
        root = Node(5)
        root.left = Node(5)
        self.assertFalse(is_valid_bst(root))

    def test_negative_values(self):
        root = Node(0)
        root.left = Node(-10)
        root.right = Node(10)
        root.left.right = Node(-5)
        self.assertTrue(is_valid_bst(root))


class TestBalancedBinaryTree(unittest.TestCase):
    def test_balanced_tree(self):
        root = Node(8)
        root.left = Node(4)
        root.right = Node(12)
        root.left.left = Node(2)
        root.left.right = Node(6)
        root.right.left = Node(10)
        root.right.right = Node(14)
        self.assertTrue(is_balanced(root))

    def test_unbalanced_tree(self):
        root = Node(8)
        root.left = Node(4)
        root.left.left = Node(2)
        root.left.left.left = Node(1)
        self.assertFalse(is_balanced(root))

    def test_empty_tree(self):
        self.assertTrue(is_balanced(None))

    def test_one_node(self):
        self.assertTrue(is_balanced(Node(10)))

    def test_difference_exactly_one(self):
        root = Node(10)
        root.left = Node(5)
        root.right = Node(15)
        root.left.left = Node(2)
        self.assertTrue(is_balanced(root))

    def test_deep_unbalance(self):
        root = Node(10)
        root.left = Node(5)
        root.right = Node(15)
        root.left.left = Node(3)
        root.left.right = Node(7)
        root.left.left.left = Node(2)
        root.left.left.left.left = Node(1)
        self.assertFalse(is_balanced(root))


suite = unittest.TestSuite()
suite.addTests(unittest.TestLoader().loadTestsFromTestCase(TestTreeTraversal))
suite.addTests(unittest.TestLoader().loadTestsFromTestCase(TestValidateBST))
suite.addTests(unittest.TestLoader().loadTestsFromTestCase(TestBalancedBinaryTree))

runner = unittest.TextTestRunner(stream=sys.stdout, verbosity=2)
result = runner.run(suite)
````

|---|---|---|

Где:
