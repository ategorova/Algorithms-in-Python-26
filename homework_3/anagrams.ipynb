# Задача 2. Anagrams
Дан список слов. Сгруппируйте слова так, чтобы в одной группе оказались все анаграммы.

## Объяснение решения

Главная идея: все анаграммы после сортировки букв становятся одинаковыми.

Например:

`eat → aet`

`tea → aet`

`ate → aet`

У всех трех слов получился одинаковый ключ `aet`.

Поэтому используем словарь, где:

- ключ — буквы слова в отсортированном порядке;
- значение — список всех слов с таким набором букв.

Для каждого слова:
1. сортируем его буквы;
2. получаем ключ;
3. добавляем слово в группу с этим ключом.

В конце возвращаем все получившиеся группы.

```python
def group_anagrams(strs: list[str]) -> list[list[str]]:
    # словарь:
    # отсортированное слово -> список его анаграмм
    groups = {}

    # проходим по всем словам
    for word in strs:
        # сортируем буквы слова
        # например eat -> aet
        key = "".join(sorted(word))

        # если такой набор букв встречаем первый раз,
        # создаем для него новую группу
        if key not in groups:
            groups[key] = []

        # добавляем слово в соответствующую группу
        groups[key].append(word)

    # возвращаем только группы слов
    return list(groups.values())
```

## Визуализация работы алгоритма

Рассмотрим:

`strs = ["eat", "tea", "tan", "ate", "nat", "bat"]`

Изначально словарь пустой:

`groups = {}`

---

### Шаг 1

Берем слово:

`eat`

Сортируем буквы:

`eat → aet`

Ключ:

`aet`

Такого ключа пока нет, поэтому создаем новую группу:

`groups = {"aet": ["eat"]}`

---

### Шаг 2

Берем:

`tea`

Сортируем:

`tea → aet`

Ключ `aet` уже есть.

Добавляем `tea` в существующую группу:

`groups = {"aet": ["eat", "tea"]}`

---

### Шаг 3

Берем:

`tan`

Сортируем:

`tan → ant`

Такого ключа еще нет.

Создаем новую группу:

`groups = {
    "aet": ["eat", "tea"],
    "ant": ["tan"]
}`

---

### Шаг 4

Берем:

`ate`

Сортируем:

`ate → aet`

Ключ `aet` уже есть.

Добавляем слово:

`groups = {
    "aet": ["eat", "tea", "ate"],
    "ant": ["tan"]
}`

---

### Шаг 5

Берем:

`nat`

Сортируем:

`nat → ant`

Ключ `ant` уже есть.

Получаем:

`groups = {
    "aet": ["eat", "tea", "ate"],
    "ant": ["tan", "nat"]
}`

---

### Шаг 6

Берем:

`bat`

Сортируем:

`bat → abt`

Такого ключа еще нет.

Итоговый словарь:

`groups = {
    "aet": ["eat", "tea", "ate"],
    "ant": ["tan", "nat"],
    "abt": ["bat"]
}`

Возвращаем только значения словаря:

`[["eat", "tea", "ate"], ["tan", "nat"], ["bat"]]`


| Слово | Отсортированные буквы | Ключ | Группа |
|---|---|---|---|
| `eat` | `aet` | `aet` | `["eat"]` |
| `tea` | `aet` | `aet` | `["eat", "tea"]` |
| `tan` | `ant` | `ant` | `["tan"]` |
| `ate` | `aet` | `aet` | `["eat", "tea", "ate"]` |
| `nat` | `ant` | `ant` | `["tan", "nat"]` |
| `bat` | `abt` | `abt` | `["bat"]` |

## Оценка сложности

Пусть:

- `n` — количество слов;
- `k` — средняя длина одного слова.

### Время: O(n · k log k)

Для каждого слова мы сортируем его буквы.

Сортировка слова длины `k` занимает:

`O(k log k)`

Таких слов `n`, поэтому общая сложность:

`O(n · k log k)`

### Память: O(n · k)

В словаре хранятся все слова и ключи для их групп.

В сумме объем хранимых данных зависит от общего количества символов во входных словах.

Поэтому дополнительная память:

`O(n · k)`

```python
import unittest
import sys


class TestGroupAnagrams(unittest.TestCase):

    def normalize(self, groups):
        # сортируем слова внутри групп
        # и сами группы, чтобы порядок не влиял на тест
        return sorted(
            [sorted(group) for group in groups]
        )

    def test_example(self):
        result = group_anagrams(
            ["eat", "tea", "tan", "ate", "nat", "bat"]
        )

        expected = [
            ["bat"],
            ["nat", "tan"],
            ["ate", "eat", "tea"]
        ]

        self.assertEqual(
            self.normalize(result),
            self.normalize(expected)
        )

    def test_one_word(self):
        result = group_anagrams(["cat"])

        expected = [["cat"]]

        self.assertEqual(
            self.normalize(result),
            self.normalize(expected)
        )

    def test_all_anagrams(self):
        result = group_anagrams(
            ["abc", "bca", "cab"]
        )

        expected = [
            ["abc", "bca", "cab"]
        ]

        self.assertEqual(
            self.normalize(result),
            self.normalize(expected)
        )

    def test_no_anagrams(self):
        result = group_anagrams(
            ["cat", "dog", "bird"]
        )

        expected = [
            ["cat"],
            ["dog"],
            ["bird"]
        ]

        self.assertEqual(
            self.normalize(result),
            self.normalize(expected)
        )

    def test_duplicates(self):
        result = group_anagrams(
            ["abc", "abc", "bca"]
        )

        expected = [
            ["abc", "abc", "bca"]
        ]

        self.assertEqual(
            self.normalize(result),
            self.normalize(expected)
        )

    def test_empty_list(self):
        result = group_anagrams([])

        self.assertEqual(result, [])


suite = unittest.TestLoader().loadTestsFromTestCase(
    TestGroupAnagrams
)

runner = unittest.TextTestRunner(
    stream=sys.stdout,
    verbosity=2
)

result = runner.run(suite)
```
