# Two sum

Дан массив целых чисел `arr` и число `k`.

Нужно найти индексы двух элементов, сумма которых равна `k`.

Гарантируется, что подходящая пара существует ровно одна.

## Объяснение решения

Проходим по массиву один раз.

Для каждого числа `number` считаем, какое второе число нужно для получения суммы `k`:

`needed = k - number`

В словаре `seen` храним числа, которые уже встретили, и их индексы.

Если `needed` уже есть в `seen`, значит нужная пара найдена:
- индекс первого числа берем из `seen`;
- индекс второго числа — текущий `index`.

Если нужного числа еще нет, сохраняем текущее число и его индекс в `seen`.

```python
def two_sum(arr: list[int], k: int):
    # здесь будем хранить:
    # число -> его индекс
    seen = {}

    # проходим по массиву
    for index, number in enumerate(arr):

        # считаем, какое число нужно добавить к текущему,
        # чтобы получить k
        needed = k - number

        # если нужное число уже встречалось,
        # значит мы нашли пару
        if needed in seen:
            return seen[needed], index

        # если пары пока нет,
        # запоминаем текущее число и его индекс
        seen[number] = index
```

```python
two_sum([1, 3, 4, 10], 7)
```

## Визуализация работы алгоритма

Рассмотрим пример:

`arr = [1, 3, 4, 10]`

`k = 7`

Нужно найти два элемента, сумма которых равна `7`.

В словаре `seen` будем хранить уже просмотренные числа и их индексы:

`число: индекс`

Изначально:

`seen = {}`

---

### Шаг 1

Текущий элемент:

`arr[0] = 1`

Ищем число, которое вместе с `1` даст `7`:

`needed = 7 - 1 = 6`

Проверяем, есть ли `6` в `seen`.

`seen = {}`

Числа `6` там нет.

Поэтому запоминаем текущий элемент:

`seen = {1: 0}`

Пару пока не нашли.

---

### Шаг 2

Переходим к следующему элементу:

`arr[1] = 3`

Считаем:

`needed = 7 - 3 = 4`

То есть теперь ищем число `4`.

Сейчас:

`seen = {1: 0}`

Числа `4` в словаре нет.

Запоминаем `3` и его индекс:

`seen = {1: 0, 3: 1}`

Пару пока не нашли.

---

### Шаг 3

Переходим к следующему элементу:

`arr[2] = 4`

Считаем:

`needed = 7 - 4 = 3`

Теперь ищем число `3`.

Смотрим в словарь:

`seen = {1: 0, 3: 1}`

Число `3` уже есть.

Его индекс:

`seen[3] = 1`

Текущий элемент `4` находится по индексу:

`2`

Получили пару:

`arr[1] + arr[2] = 3 + 4 = 7`

Поэтому возвращаем:

`1 2`

После этого алгоритм заканчивается, и элемент `10` уже проверять не нужно.

### То же самое в таблице

| Шаг | Индекс | Текущее число | `needed = k - number` | `seen` до проверки | Что делаем |
|---:|---:|---:|---:|---|---|
| 1 | 0 | 1 | 6 | `{}` | 6 нет → сохраняем `1: 0` |
| 2 | 1 | 3 | 4 | `{1: 0}` | 4 нет → сохраняем `3: 1` |
| 3 | 2 | 4 | 3 | `{1: 0, 3: 1}` | 3 есть → нашли пару |

Результат:

`1 2`

```python
import unittest
import sys


class TestTwoSum(unittest.TestCase):

    def test_example_1(self):
        self.assertEqual(
            two_sum([1, 3, 4, 10], 7),
            (1, 2)
        )

    def test_example_2(self):
        self.assertEqual(
            two_sum([5, 5, 1, 4], 10),
            (0, 1)
        )

    def test_pair_at_edges(self):
        self.assertEqual(
            two_sum([2, 7, 11, 15], 17),
            (0, 3)
        )

    def test_negative_numbers(self):
        self.assertEqual(
            two_sum([-5, 2, 7, 10], 5),
            (0, 3)
        )

    def test_zero(self):
        self.assertEqual(
            two_sum([0, 4, 8], 8),
            (0, 2)
        )

    def test_two_elements(self):
        self.assertEqual(
            two_sum([3, 6], 9),
            (0, 1)
        )

    def test_negative_target(self):
        self.assertEqual(
            two_sum([-7, -3, 2, 5], -10),
            (0, 1)
        )


suite = unittest.TestLoader().loadTestsFromTestCase(TestTwoSum)

runner = unittest.TextTestRunner(
    stream=sys.stdout,
    verbosity=2
)

result = runner.run(suite)
```

## Оценка сложности

### Время: O(n)

Мы проходим по массиву только один раз.

Для каждого элемента:
- вычисляем `needed`;
- проверяем его наличие в словаре;
- при необходимости добавляем текущий элемент в словарь.

Поиск и добавление элемента в словарь в среднем работают за `O(1)`.

Поэтому для массива из `n` элементов общая временная сложность:

**O(n)**

### Память: O(n)

В худшем случае перед нахождением пары в словаре `seen` может находиться почти весь массив.

Поэтому дополнительная память:

**O(n)**
